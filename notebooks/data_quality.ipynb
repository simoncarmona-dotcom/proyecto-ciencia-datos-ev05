### Respuestas a las preguntas de análisis (Momento 3)

* **¿Qué diferencias existen entre `int64`, `float64` y `object`?**
  * `int64`: Representa números enteros de 64 bits sin parte decimal (por ejemplo, identificadores como `order_id`, conteos como `order_number` o indicadores binarios como `reordered`).
  * `float64`: Representa números de punto flotante (decimales) de 64 bits. En Pandas, también se asigna automáticamente a columnas numéricas que contienen valores nulos (`NaN`), como ocurre en `days_since_prior_order` y `add_to_cart_order`.
  * `object`: Representa cadenas de texto (`strings`) o datos mixtos en Pandas, como los nombres de productos (`product_name`), pasillos (`aisle`) y departamentos (`department`).
* **¿Qué información proporciona `info()` que no se obtiene directamente con `shape`?**
  Mientras que `shape` únicamente devuelve una tupla con el total de filas y columnas `(filas, columnas)`, el método `info()` detalla el tipo de índice (`RangeIndex`), los nombres de todas las columnas, el **conteo de valores no nulos (`Non-Null Count`)** por cada columna (clave para detectar datos faltantes), el tipo de dato (`Dtype`) de cada variable y la cantidad de memoria RAM utilizada por el DataFrame.
* **¿En qué columnas se observan valores ausentes?**
  Se observan valores ausentes únicamente en tres columnas:
  1. `days_since_prior_order` del DataFrame `orders` (28.819 valores ausentes).
  2. `product_name` del DataFrame `products` (1.258 valores ausentes).
  3. `add_to_cart_order` del DataFrame `order_products` (836 valores ausentes).
* **¿Qué posibles dificultades podrían generar estos valores en un análisis posterior?**
  Los valores ausentes pueden provocar diferencias en los conteos de registros frente al total de filas, alterar cálculos si se reemplazan incorrectamente por cero, impedir la conversión directa de columnas `float64` a `int64`, dificultar la identificación por nombre de los 1.258 productos sin `product_name` al generar reportes comerciales, o excluir registros automáticamente al aplicar filtros lógicos sobre las columnas afectadas.

### Filtros con condiciones simultáneas

#### Condición simultánea 1: Pedidos realizados a las 10 horas y con `order_number` superior a 10 (`orders`)

```python
orders.loc[(orders['order_hour_of_day'] == 10) & (orders['order_number'] > 10)]
```

**Explicación del resultado:** Al exigir simultáneamente que el pedido se haya realizado a las 10 horas (`40.578` pedidos) **y** que su `order_number` sea mayor a 10 (`244.072` pedidos), el subconjunto se reduce a **21.537 pedidos** que cumplen ambas condiciones.

#### Condición simultánea 2: Productos que pertenezcan al departamento 1 y tengan un `aisle_id` superior a 50 (`products`)

```python
products.loc[(products['department_id'] == 1) & (products['aisle_id'] > 50)]
```

**Explicación del resultado:** De los 4.008 productos que pertenecen al departamento 1, únicamente **1.618 productos** están ubicados además en un pasillo con identificador superior a 50 (`aisle_id > 50`).

### Respuestas a las preguntas de análisis (Momento 4)

* **¿Qué diferencia existe entre seleccionar columnas y filtrar filas?**
  Seleccionar columnas (proyección vertical) reduce la cantidad de variables o atributos visibles manteniendo intacta la cantidad de registros (por ejemplo, pasar de 4 a 2 columnas en `products` conservando las 49.694 filas). En cambio, filtrar filas (selección horizontal) evalúa una condición booleana sobre cada registro y devuelve únicamente el subconjunto de filas que cumplen el criterio, manteniendo las columnas originales.
* **¿Qué ventajas ofrece `loc[]` al consultar información específica?**
  El indexador `loc[filas, columnas]` permite filtrar filas mediante condiciones booleanas y, en la misma instrucción, seleccionar explícitamente por etiqueta qué columnas se desean conservar, haciendo el código más legible, explícito y seguro frente a advertencias de indexación encadenada en Pandas.
* **¿Qué ocurre con la cantidad de registros cuando se aplican varias condiciones?**
  Cuando se combinan condiciones mediante el operador de intersección lógica `&` (`AND`), el criterio se vuelve más restrictivo y la cantidad de registros disminuye o se mantiene igual, pero nunca supera a la de las condiciones individuales (por ejemplo, de 40.578 pedidos a las 10 h se pasó a 21.537 pedidos al exigir también `order_number > 10`). Si se usara el operador `|` (`OR`), la cantidad de registros aumentaría o se mantendría igual al unir ambos subconjuntos.
* **¿Por qué es importante utilizar correctamente los operadores lógicos?**
  En Pandas es indispensable utilizar los operadores vectorizados elemento a elemento (`&` para AND, `|` para OR, `~` para NOT) y encerrar cada condición individual entre paréntesis `(...)`. Si se usan las palabras clave de Python `and` / `or` o se omiten los paréntesis, Python lanza un error (`ValueError` o error de precedencia de operadores) o evalúa incorrectamente las series booleanas.
